```python
def get_device_list(string,device,geometry_files):
    device_list = []
    for ifile in geometry_files:
        if string in ifile:
            with open(ifile, 'r') as f:
                data = json.load(f)
            for idev in data[0]["devices"]:
                if device in idev["production_id"]:
                    device_list.append(idev["production_id"])
    return device_list
```

```python
def prod_id_to_icm_id(prod_id):
    '''
    production id
    '''
    icm_id_list = []
    for ifile in geometry_files:
        with open(ifile, 'r') as f:
            data = json.load(f)
        for idev in data[0]["devices"]:
            if idev["production_id"] == prod_id:
                icm_id_list.append(idev["icm_id"])
    if len(icm_id_list)>1:
        print(f"more than one icm id found for prod id {prod_id} {icm_id_list}")
    return icm_id_list[0]
```

```python
written_degg_list = []
written_pmt_list = []
degg_dict = {}
for idegg in deployed_device_list[:]:
    icm_id = prod_id_to_icm_id(idegg)
    fat_tt_folder = [idir for idir in degg_list if idegg in idir]
    if len(fat_tt_folder) == 0:
        print(f"missing {idegg}")
        degg_dict[idegg] = {"icm_id": icm_id, "transit_times": {}}
    elif len(fat_tt_folder)> 0:
        written_degg_list.append(idegg)
        fat_tt_folder = fat_tt_folder[0]
        meas_files = glob.glob(fat_tt_folder+"/*")
        pmt_list = list(set([extract_pmt(ifile) for ifile in meas_files]))
        if len(pmt_list) != 2:
            print(f" {idegg} missing PMTs?############################### {pmt_list}")
        pmt_dict = {}
        for ipmt in pmt_list:
            meas_files_this_pmt = [ifile for ifile in meas_files if ipmt in ifile]
            meas_temperature = [extract_temperature(ifile) for ifile in meas_files_this_pmt]
            tt_info_map = []
            if len(meas_temperature) > 0:
                min_temp = min(meas_temperature)
                # print(min_temp)
                meas_file_min_temp = [ifile for ifile in meas_files_this_pmt if abs(extract_temperature(ifile)-min_temp) < 0.01][0]
                a,mu,sigma,chi2,pvalue = extract_fit_params(meas_file_min_temp)
                if np.isnan(chi2):
                    meas_temperature = [extract_temperature(ifile) for ifile in meas_files_this_pmt if ifile not in [meas_file_min_temp]]
                    min_temp = min(meas_temperature)
                    meas_file_min_temp = [ifile for ifile in meas_files_this_pmt if abs(extract_temperature(ifile)-min_temp) < 0.01][0]
                    a,mu,sigma,chi2,pvalue = extract_fit_params(meas_file_min_temp)

                pmt_name = extract_pmt(meas_file_min_temp)
                written_pmt_list.append(idegg+ipmt)

                # print(pmt_name)
                time = extract_time(meas_file_min_temp)
                time = str(pd.to_datetime(time, unit='s').strftime('%Y%m%d'))
                # print(f"time{time}")
                run_number = extract_run_number(meas_file_min_temp)
                temp = extract_temperature(meas_file_min_temp)
                # print(meas_file_min_temp)
                # hv = [ielt["voltage"] for ielt in hv_data if ielt["pmt_name"]==ipmt]
                # hv = [ielt["meas_date"] for ielt in hv_data if ielt["pmt_name"]==ipmt]
                hv_list = [float(ielt["voltage"][:-1]) for ielt in hv_data if ielt["pmt_name"]==pmt_name and str(ielt["meas_date"]) == str(time)]
                if len(hv_list) == 1:
                    hv = hv_list[0]
                elif len(hv_list) > 1:
                    print(f"multiple hvs, some issues")
                # print(hv)
                tt_info = {"run_number": run_number, "sigma": sigma,"a": a, "b": mu, "c": sigma, "chi2": chi2,"applied HV": hv, "temperature": temp}
                tt_info_map.append(tt_info)
            pmt_dict[f"pmt_{ipmt}"] = tt_info_map
        degg_dict[idegg] = {"icm_id": icm_id, "transit_times": pmt_dict}

print(f"written degg list {len(written_degg_list)}")
print(f"written degg list {len(written_pmt_list)}")

            


```
